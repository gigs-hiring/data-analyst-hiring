# 1. How much data does a subscription typically consume?

### Query

```python
query_avg_usage = """

with plan_usage_per_subscription as (
    select
          ud.subscription_id
        , p.organization_name
        , sum(ud.cumulative_data_usage_megabyte / 1000) as total_data_usage
        , (avg(ud.cumulative_data_usage_megabyte) / 1000)  as avg_data_usage_gb
    from usage_data ud
    
    left join projects p
        on ud.project_id__hashed = p.project_id__hashed

    where p.project_type = 'Connect'
    
    group by all
),

avg_monthly_usage as (   
    select
          'Avg. Monthly Usage (GB)' as segment
        , organization_name
        , avg(avg_data_usage_gb::decimal(18,1)) as avg_monthly_data_use
    from plan_usage_per_subscription
    
    where avg_data_usage_gb > 0
    
    group by all order by 1 asc
)

select * from avg_monthly_usage

"""

df_avg_usage = sql_to_df(query_avg_usage)
df_avg_usage
```

### Visualisation

```python
# --- aggregate -> one value per (segment, organization) ---
agg = df_avg_usage.groupby(["segment", "organization_name"], as_index=False).agg(avg_monthly_data_use=("avg_monthly_data_use", "mean"))


# Preserve the order of segments as they first appear
seg_order = list(df_avg_usage["segment"].drop_duplicates())


# Consistent org order (alphabetical; change if you prefer another order)
org_order = sorted(agg["organization_name"].unique())


# Pivot: rows=segment, columns=organization, values=mean consumption
pv = agg.pivot(index="segment", columns="organization_name", values="avg_monthly_data_use").reindex(seg_order).reindex(columns=org_order)


# --- Plot grouped barh ---
fig, ax = plt.subplots(figsize=(15, 2))
pv.plot(kind="barh", ax=ax, legend=True)   # grouped by default

# Axes, labels, limits
ax.set_xlabel("Avg. Data Consumption (GB)")
ax.set_title("# Avg. Data Usage (GB)")
ax.grid(axis="x", alpha=0.3)
ax.set_xlim(0, max(4, pv.max().max() * 1.15))  # keep your old cap but allow larger if needed
ax.legend(title="Organization", frameon=False, ncol=1)

# Value labels on each bar
for p in ax.patches:
    w = p.get_width()
    if pd.notna(w):
        ax.annotate(f"{w:.2f} GB",
                    (w, p.get_y() + p.get_height()/2),
                    xytext=(6, 0), textcoords="offset points",
                    va="center", ha="left", fontsize=9)

plt.subplots_adjust(top=0.90)
plt.tight_layout()
plt.show()
```

## Quick Findings

- Higher average use for SmartDevices Inc. compared with ACME Phones
- Possibly due to the distribution and usage of plans

# 2. How does usage look like at different plan data allowances?

### Query

```python
sql_usage_by_plan_allowance = """

with data_allowance_per_plan as (
    select distinct
          plan_id
        , case when data_allowance_mb is null then 'Unlimited' else (cast(round(data_allowance_mb/1000,0) as varchar) || ' GB') end as plan_allowance_name_gb
        , data_allowance_mb/1000 as plan_allowance_value_gb
    from plan_events
),

usage_data_enriched as (
    select
          ud.subscription_id
        , ud.reporting_date
        , p.organization_name
        , ud.cumulative_data_usage_megabyte
        , dapp.plan_allowance_name_gb
        , dapp.plan_allowance_value_gb
    from usage_data ud
    
    left join data_allowance_per_plan dapp
        on ud.plan_id = dapp.plan_id
        
    left join projects p
        on ud.project_id__hashed = p.project_id__hashed

    where p.project_type = 'Connect'
),

usage_data_final as (
    select
          plan_allowance_name_gb
        , organization_name
        , count(distinct subscription_id) as subscriptions
        , avg(cumulative_data_usage_megabyte / 1000) as avg_monthly_usage_gb
        , sum(cumulative_data_usage_megabyte / 1000) / count(distinct subscription_id) as total_avg_usage_gb
    from usage_data_enriched
    group by all 
)

select * from usage_data_final order by 3 desc

"""

df_usage_by_plan = sql_to_df(sql_usage_by_plan_allowance)
df_usage_by_plan.head(5)
```

### Visualisation

```python
import matplotlib.pyplot as plt
import pandas as pd

# --- 1) Aggregate by plan + organization ---
order = ["Unlimited", "5.0 GB", "1.0 GB"]  # adjust if your labels differ

g_org = df_usage_by_plan.groupby(["plan_allowance_name_gb", "organization_name"], as_index=False).agg(monthly=("avg_monthly_usage_gb", "mean"))


org_order = sorted(g_org["organization_name"].unique())

# --- 2) Pivot to get one series per org ---
monthly_pv = (
    g_org.pivot(index="plan_allowance_name_gb", columns="organization_name", values="monthly")
        .reindex(order)
        .reindex(columns=org_order)
)

# --- 3) Plot single grouped barh chart ---
fig, ax = plt.subplots(figsize=(10, 3))
monthly_pv.plot(kind="barh", ax=ax, legend=True)  # use stacked=True if you prefer stacked

ax.set_title("Avg. Monthly Usage by Plan")
ax.set_xlabel("Avg usage (GB)")
ax.set_ylabel("Plan allowance")
ax.grid(axis="x", alpha=0.5)
ax.invert_yaxis()  # first item at the top
ax.set_xlim(0, monthly_pv.max().max() * 1.15)
ax.legend(title="Organization", frameon=False)

# --- value labels on bars ---
for p in ax.patches:
    w = p.get_width()
    if pd.notna(w):
        ax.annotate(f"{w:.2f} GB",
                    (w, p.get_y() + p.get_height()/2),
                    xytext=(6, 0), textcoords="offset points",
                    va="center", ha="left", fontsize=9)

plt.tight_layout()
plt.show()
```

#### Quick observations & Findings

- Overall data usage is somewhat expected. Where usage Unlimited > 5GB > 1GB
- We actually see higher usage for ACME Phone compared with SmartDevices Inc
- This is driven by the significant number of users on the 1GB plan for ACME, which pulls down the overall average.

# 3. Do subscriptions typically consume consistent amounts of data throughout their lifetime?

### Query

```python
sql_usage_by_plan_allowance = """

with data_allowance_per_plan as (
    select distinct
          plan_id
        , case when data_allowance_mb is null then 'Unlimited' else (cast(data_allowance_mb/1000 as varchar) || 'GB') end as plan_allowance_name_gb
        , data_allowance_mb/1000 as plan_allowance_value_gb
    from plan_events
),

usage_data_enriched as (
    select
          ud.subscription_id
        , p.organization_name
        , ud.plan_id
        , ud.subscription_period_number
        , ud.project_id__hashed
        , ud.reporting_date
        , ud.subscription_period_start
        , ud.subscription_period_end
        , ud.cumulative_data_usage_megabyte
        , dapp.plan_allowance_name_gb
        , dapp.plan_allowance_value_gb
    from usage_data ud
    
    left join data_allowance_per_plan dapp
        on ud.plan_id = dapp.plan_id

    left join projects p
        on ud.project_id__hashed = p.project_id__hashed
        
    where p.project_type = 'Connect'
),

usage_data_final as (
    select
          plan_allowance_name_gb
        , subscription_period_number
        , organization_name
        , count(distinct subscription_id) as subscriptions
        , avg(cumulative_data_usage_megabyte / 1000) as avg_usage_gb
        , sum(cumulative_data_usage_megabyte / 1000) as total_usage_gb
        , count(reporting_date) as active_months
    from usage_data_enriched
    group by all
)

select * from usage_data_final order by 3 desc
"""

df = sql_to_df(sql_usage_by_plan_allowance)
df.head(20)
```

### Visualisation

```python
# --- pick the exact labels used in your data ---
plan_order = ["1.0GB", "5.0GB", "Unlimited"]   # change to ["1.0 GB","5.0 GB","Unlimited"] if your labels include spaces

# All subscription periods you want on the x-axis
periods = range(1, int(df["subscription_period_number"].max()) + 1)

fig, axes = plt.subplots(1, len(plan_order), figsize=(25, 5), sharex=True, sharey=False, constrained_layout=False)
axes = axes.ravel()

for ax, plan in zip(axes, plan_order):
    # ---- aggregate + pivot to have one column per organization ----
    d = (
        df[df["plan_allowance_name_gb"] == plan]
        .groupby(["subscription_period_number", "organization_name"], as_index=False)
        .agg(avg_usage=("avg_usage_gb", "mean"))          # mean usage per period/org
    )

    pv = (
        d.pivot(index="subscription_period_number", columns="organization_name", values="avg_usage")
         .reindex(periods)                                 # ensure a full 1..N index
    )

    # ---- plot each organization as its own line ----
    for org in pv.columns:
        ax.plot(pv.index, pv[org], linewidth=2, marker='o', markevery=1, label=org)

        # annotate all points (skip NaNs)
        for xi, yi in zip(pv.index, pv[org]):
            if pd.notna(yi):
                ax.annotate(f"{yi:,.2f}",
                            (xi, yi),
                            textcoords="offset points", xytext=(0, 9),
                            ha="center", fontsize=8,
                            bbox=dict(boxstyle="round,pad=0.15", fc="white", alpha=0.7, lw=0))

    # axis cosmetics (kept from your original)
    ax.set_title(plan)
    ax.set_xlabel("Subscription period")
    ax.set_ylabel("Avg usage (GB)")
    ax.grid(axis='y', alpha=0.3)

    # vertical breathing room
    vals = pv.to_numpy().ravel()
    vals = vals[np.isfinite(vals)]
    if vals.size:
        vmin, vmax = vals.min(), vals.max()
        pad = 0.12 * (vmax - vmin if vmax > vmin else 1.0)
        ax.set_ylim(max(0, vmin - pad), vmax + pad)

    ax.margins(x=0.03, y=0.15)
    ax.legend(title="Organization", frameon=False, loc="best")

# overall layout
for ax in axes:
    ax.set_ylim(bottom=0)

fig.subplots_adjust(hspace=0.25, wspace=0.12)
fig.suptitle("Average Usage by Plan Allowance (per plan, by organization)", y=0.98, fontsize=16)
plt.show()
```

### Quick Findings

- Generally it looks like usage is only stable over time for the 5GB plan
- The unlimited plan data is somewhat misleading. It looks decreasing since we're drawn to the 4-month peak of 11.94 after which the data decreases
- However, if we didn't have these peaks, month 1 to month 12 would actually look pretty stable
- Further analysis required to identify this initial spike as a possible outlier with extremely high usage

# 4. Compare the retention pattern for the most recently launched project versus the two older ones.

### Query

```python
## projects

query = """
    with subscriptions_enriched as (
        select
              subscription_id
            , project_type
            , device_type
            , subscription_period_number
            , reporting_date
            , subscription_period_start
            , organization_name
            , first_value(subscription_period_start) over (partition by subscription_id order by subscription_period_start asc) as subscription_start_date
        from usage_data ud
        left join projects p
            on ud.project_id__hashed = p.project_id__hashed
        --where device_type = 'Phones'
    ),
    
    subscriptions_enriched_agg as (
        select
              date_trunc('month', subscription_start_date) as start_cohort
            , organization_name
            , project_type
            , device_type
            , concat(project_type, ' (' , organization_name, ')' , ': ', device_type) as segment
            , count(distinct case when subscription_period_number = 1 then subscription_id end) as new_subscriptions
            , count(distinct case when subscription_period_number = 2 then subscription_id end) as subscriptions_retained_m1
            , round(count(distinct case when subscription_period_number = 2 then subscription_id end) / count(distinct case when subscription_period_number = 1 then subscription_id end) , 4) as retained_m1_pct
        from subscriptions_enriched
        where subscription_start_date >= '2025-04-01' and subscription_start_date < '2025-05-01'
        group by all
    )
    
    select * from subscriptions_enriched_agg order by 1,2

"""

df = sql_to_df(query)
df
```

### Visualisation

```python
relevant_segments = ['API (People Mobile): Phones', 'Connect (ACME Phone): Phones', 'Connect (SmartDevices Inc.): Wearables']

plotdf = df[df['segment'].isin(relevant_segments)][['segment','retained_m1_pct']]


# enforce order
plotdf['segment'] = pd.Categorical(plotdf['segment'], categories=relevant_segments, ordered=True)
plotdf = plotdf.sort_values('segment')


# horizontal bar chart (values are 0–1 fractions)
ax = plotdf.set_index('segment')['retained_m1_pct'].plot(kind='barh', figsize=(10,3), legend=False)


ax.set_title('Retention M1 [%]: API vs Connect: M1 Retention [%]')
ax.set_xlabel('Retention (M1)')
ax.xaxis.set_major_formatter(StrMethodFormatter('{x:.0%}'))  # show as percentages
ax.set_xlim(0, 0.6)


# value labels on bars
for p in ax.patches:
    w = p.get_width()
    ax.annotate(f'{w:.1%}', (w, p.get_y() + p.get_height()/2),
                xytext=(6, 0), textcoords='offset points',
                va='center', ha='left', fontsize=9)


ax.invert_yaxis()  # show first item at the top
ax.grid(axis='x', alpha=0.3)
plt.show()
```

### Quick Findings
- Older cohorts have significantly higher retention rates than more recent ones. Would make sense to dive into this.
- Retention Rate is significantly lower for the API product, compared with other products.
- Are these products comparable? Are prior retention rates the best point of comparison to evalute initial success of API
